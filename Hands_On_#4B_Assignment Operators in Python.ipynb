```python
#Student Score Tracker - Question 1

#Variables:

#Initial score
score = 60

#Add 15 marks for class assignment
score += 15

#Subtract 5 marks for late submission  
score -= 5

#Double the score for special bonus
score *= 2

#Print final score
Final_Score = score
print(Final_Score)
```

```python
#Digital Wallet Simulator - Question 2

#Variables:

#Initial balance
balance = 25000

#Deposit #8,500
balance += 8500

#Withdraw #4,200
balance -= 4200

#Divide equally among 3 savings accounts
balance /= 3

#Print final balance
Final_Balance = balance
print(Final_Balance)

#Print data type of balance
print(type(balance))
```

```python
#Inventory Management System - Question 3

#Variables:

#Assign Named stock
stock = 150

#Customer Purchase
stock -= 28

#New Shipment Arrival
stock += 75

#Pack into carton of 12 items each
stock //= 12

#Print the number of full cartons
full_cartons = stock
print(full_cartons)
```
