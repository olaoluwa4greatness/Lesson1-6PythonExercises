```python
#Exercise 2 - Payroll System

#Variables:
Employee_Name = "Ali Edet"
Department = "Logistics"
Basic_Salary = 255000.000
Bonus = 23000.000
Tax_Rate = 7.5/100

#Calculated Output:
Gross_Salary = Basic_Salary + Bonus
Tax_Amount = Gross_Salary * Tax_Rate
Net_Salary = Gross_Salary - Tax_Amount

#Using F-strings:
Payroll_Summary = ( F" Employee: {Employee_Name}. "
                   F" Department: {Department}. "
                   F" Basic Salary: #{Basic_Salary}. "
                   F" Bonus: #{Bonus}. "
                   F" Gross Salary: #{Gross_Salary}. "
                   F" Tax: #{Tax_Amount}. "
                   F" Net Salary #{Net_Salary}. ")

print(Payroll_Summary)
```

```python
#Exercise 3 - E-commerce Order Receipt

#Variables:
Customer_Name = "Amina Bello"
Product = "Wireless Headphones"
Quantity = 3
Unit_Price = 27500

#Calulated Output:
Total_Cost = 3 * 27500

#Challenge
#Since Quantity is greater than 3
Discount = Total_Cost * 10/100
Final_Total_Cost = Total_Cost - Discount


#Using F-string:
Online_Store_Receipt = ( F" Customer: {Customer_Name}. "
                         F" Product: {Product}. "
                         F" Quantity: {Quantity}. "
                         F" Unit Price: {Unit_Price}. "
                         F" Total Cost: {Final_Total_Cost}. "
                         F" Thank you for shopping with us! " )


#Output:
print(Online_Store_Receipt)
```
