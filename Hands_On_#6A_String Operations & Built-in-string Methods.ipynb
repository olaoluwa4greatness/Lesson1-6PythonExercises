--Hands-on Exercise 2 - Fraud Detection Log Analyzer (Banking & Cybersecurity Industry)--

```python
#Task

#print original log
Original_Log = "   LOG SUCCESSFUL FROM ATM-045 ON 2026-07-25   "
print(Original_Log)

#remove spaces
print(Original_Log.strip())

#convert cleaned log to lowercase
print(Original_Log.lower())

#find position of word
print(Original_Log.find("SUCCESSFUL"))

#replace word with another word
print(Original_Log.replace("SUCCESSFUL","VERIFIED"))
```

------BONUS CHALLENGE-----

```python
#number of characters before cleaning

Original_Log = "   LOG SUCCESSFUL FROM ATM-045 ON 2026-07-25   "

def count_characters(text):
    return len(text)
print(count_characters(Original_Log))

#number of characters after cleaning
print(count_characters("LOG SUCCESSFUL FROM ATM-045 ON 2026-07-25"))
```

Hands-on Exercise 3 - Movie Recommendation Search Engine (Entertainment & AI Industry)

```python
#Task

#Variable
Movie_Title = "The Last Kingdom Returns"

#print first character
print(Movie_Title[0])

#print last character
print(Movie_Title[-1])

#print first word only
print(Movie_Title[0:3])

#print last word only
print(Movie_Title[17:24])

#find position of 'Returns'
print(Movie_Title.find("Returns"))

#split the movie title into individuals
print(Movie_Title.split())
```

```python
#display total number of characaters

Movie_Title = "The Last Kingdom Returns"

def count_characters(text):
    return len(text)
print(count_characters(Movie_Title))

#coverting title to title()
print(Movie_Title.title())
```
